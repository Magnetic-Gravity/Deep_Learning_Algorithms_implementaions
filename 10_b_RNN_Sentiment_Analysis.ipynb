- Recurrent Neural Networks :  RNNs are deep learning models designed to process sequential data, such as text, by using memory to retain context from previous inputs.
- Sentiment analysis uses these models to automatically classify text (reviews, tweets) as positive, negative, or neutral, understanding the context to identify opinions

```python
import numpy as np

docs = [
    'go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad'
]
```

Tokenization is a fundamental step in Natural Language Processing (NLP). It involves dividing a Textual input into smaller units known as tokens. These tokens can be in the form of words, characters, sub-words or sentences.

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='moin')
```

```python
tokenizer.fit_on_texts(docs) # word by word saparation, index_generator, will change capital letter words in small letter words
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count # How many sentence are there
```

```python
  sequences = tokenizer.texts_to_sequences(docs) # which word is in what place(index) in a sentence
sequences
```

# Padding  
- it is a preprocessing technique that standardizes input sequence lengths by adding special tokens—usually zeros—to shorter sentences, ensuring all data in a batch is of uniform size.
- It enables efficient tensor computation on GPUs/TPUs, allowing models like LSTMs and CNNs to process variable-length text as fixed-size batches.

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences, padding='post')
sequences
```

# Apply Sentiment Analysis on IMDB dataset

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
print(X_train) # Already Tokenized data and also integer encoded
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(X_train[0])
```

```python
len(X_train[0]) # len of a single review in the dataset
```

```python
# Every review will consist of only 50 words
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(X_train[0]) # 1st Review consisting of 50 words
```

```python
model = Sequential()

# if we take return_sequences true, then the output will go to outside as well as to the next timesteps (Named Entity Recognition)
model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False)) # 50(timesteps),1(input),32(RNN layer nodes), output go only next
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5, validation_data=(X_test,y_test))
```

# Embedding
  - Semantic Representation: Unlike sparse one-hot encoding, embeddings capture the "meaning" of data, allowing models to understand that "king" and "queen" are related.
  
  - Dimensionality Reduction: They convert high-dimensional, categorical data into smaller, manageable vectors, improving computational efficiency.

```python
import numpy as np

docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences, padding='post')
sequences
```

```python
model = Sequential()
model.add(Embedding(17,output_dim=2,input_shape=(5,))) #  unique_words(17),embedding_dense=2(represents every word with 2 numbers)

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

# Embedding on IMDB dataset

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train.shape
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
model = Sequential()

model.add(Embedding(10000, 2, input_shape=(50,))) # vocublary_size=1000, for every word a dense vector of dims 2
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```
