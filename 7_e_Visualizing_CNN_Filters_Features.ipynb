# VGG16 CNN model Visualization
- It is commonly used for image classification tasks.
- It developed by the Visual Geometry Group at Oxford, is renowned for its uniform architecture consisting of 16 layers (13 convolutional and 3 fully connected) that use 3x3 filters, max-pooling, and ReLU activation with input size (224,224,3).

```python
from keras.applications.vgg16 import VGG16

# Load the model
model = VGG16()
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model)
```

```python
for i in range(len(model.layers)):
  #check for convolutional layer
  if 'conv' not in model.layers[i].name:
    continue
  #get filter weights
  filters, biases = model.layers[i].get_weights()
  print('Layer', i, model.layers[i].name, filters.shape)
```

```python
# retrieve weights from the second hidden layer
filters, biases = model.layers[1].get_weights()
```

```python
# Normalize filter values to 0-1 so we can visualize them
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as plt
n_filters = 6
ix = 1   # Keeps track of subplot index (position in grid)
fig = plt.figure(figsize=(10,10))
for i in range(n_filters):
  #get the filters
  f = filters[:,:,:,i] # filters is assumed to be a 4D tensor (height, width, channels, number_of_filters)
  for j in range(3):
    #subplot for 6 filters and 3 channels
    plt.subplot(n_filters, 3, ix)  # rows=n_filters(6), cols=3(channels), ix decides where to place the image
    plt.imshow(f[:,:,j], cmap='gray') # f[:,:,j] selects one channel of the filter
    ix+=1

plt.show()
```

```python
from keras.models import Model
model = Model(inputs=model.inputs, outputs=model.layers[1].output)
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from numpy import expand_dims
from tensorflow.keras.applications.vgg16 import preprocess_input
```

```python
image = load_img('/content/virat_kohli.jpg', target_size=(224,224))
image = img_to_array(image) # convert image to an array
image = expand_dims(image, axis=0) # expand dimensions so that it represents a single 'sample'
image = preprocess_input(image)
```

```python
features = model.predict(image)
fig = plt.figure(figsize=(20,15))

for i in range(1, features.shape[3]+1):
  plt.subplot(8,8,i)
  plt.imshow(features[0,:,:,i-1],cmap='gray')

plt.show()
```

```python
model2 = VGG16()
```

```python
layer_index = [2,5,9,13,17]
outputs = [model2.layers[i].output for i in layer_index]
model3 = Model(inputs=model2.inputs, outputs=outputs)
```

```python
feature_map = model3.predict(image)

for i, fmap in zip(layer_index, feature_map):
  fig = plt.figure(figsize=(20,15))
  fig.suptitle("Layer_{}".format(i), fontsize=20)
  for i in range(1,features.shape[3]+1):
    plt.subplot(8,8,i)
    plt.imshow(fmap[0,:,:,i-1], cmap='gray')
plt.show()
```
