# Data augmentation 
- It increases dataset size and diversity by applying transformations (rotation, flipping, zooming, color adjustments) to existing data without changing labels. It acts as a regularization technique to prevent overfitting, improve model generalizability, and enhance robustness against noise. Key techniques include geometric transformations, color space modifications, and synthetic data generation, widely used in computer vision and NLP.

```python
import tensorflow 
from tensorflow import keras
from keras.preprocessing import image

img = image.load_img('augmentation/dog.jpg', target_size=(200, 200))
```

```python
img
```

```python
type(img) #PIL->Python image library
```

```python
# Transformation that we will apply
import tensorflow as tf
datagen = tf.keras.preprocessing.image.ImageDataGenerator(
    rotation_range=30,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2
)
```

```python
img = image.img_to_array(img) # this will convert image into numpy array
type(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
# Save the images into aug folder
# flow is used for single image and flow_from_directory is used for multiple images
i=0
for output in datagen.flow(input_batch, batch_size=1, save_to_dir='augmentation/aug'):
    i=i+1
    if i==10:
        break
```
