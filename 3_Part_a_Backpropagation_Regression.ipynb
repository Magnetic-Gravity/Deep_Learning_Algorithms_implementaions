```python
#layer_dims = [2(input_neuron), 2(hidden_neuron), 1(output_neuron)]  - List containing the number of neurons in each layer
def initialize_parameters(layer_dims):
  np.random.seed(3)    # Every run gives same weights → easier debugging, Fixes randomness so results are reproducible
  parameters = {}      # Empty dictionary to store: [[W1, b1] ,[W2, b2]]
  L = len(layer_dims)  # Number of layers (including input layer) L = 3

  for l in range(1,L): # Why start from 1? -> Layer 0 is input layer → no weights, We creates weight between layer1 and layer0
    # Creates keys: W1, W2, np.ones((rows, cols)) and matrix of 1's, np.ones(neurons in previous layer, neurons in current layer)
    parameters['W'+str(l)] = np.ones((layer_dims[l-1],layer_dims[l]))*0.1  # Why 0.1 -> Prevents large activations, weight scaling
    parameters['b'+str(l)] = np.zeros((layer_dims[l],1))  # Bias is added to each neuron, Shape: (neurons_in_layer, 1)
  return parameters
```

```python
initialize_parameters([2,2,1])
```

```python
# If First layer, A_prev = input features X and  If Hidden layer, A_prev = output of previous activation
def linear_forward(A_prev, W, b):
  Z = np.dot(W.T, A_prev) + b
  return Z  # Z is not the final output, just the linear part.
```

```python
#Forward Propagation -> data flows from input → hidden → output
def L_layer_forward(X, parameters): # X-input data (features), parameters-List containing all weights and biases
  A = X
  L = len(parameters) // 2     # Parameters [[W1,b1],[W2,b2]] which means 4//2 = 2

  for l in range(1, L+1):  # Iterates layer by layer, Forward Propagation always uses A(l) = f( W(l)ᵀ.A(l−1)) + b(l)
    A_prev = A             # save the previous layer’s output  initially A_prev = A0 = X
    W1 = parameters['W' + str(l)]  # Fetches the weight matrix for layer l
    b1 = parameters['b' + str(l)]  # Fetches bias vector for layer l
    # print("A" + str(l-1) + " : ", A_prev)
    # print("W" + str(l) + " : ", W1)  # weights of the current layer
    # print("b" + str(l) + " : ", b1)  # bias vector for the current layer
    # print("--"*20)   # Makes output readable when many layers are printed

    A = linear_forward(A_prev, W1, b1)
    # print("A" + str(l) + " : ", A) # Output of current layer
    # print("**"*20)  # readability

  return A, A_prev   # A output of the current network, A_prev previous layer’s activation
```

```python
def update_parameters(parameters,y,y_hat,A1,X):
  # Output Layer
  parameters['W2'][0][0] += (0.001 * 2 * (y-y_hat)*A1[0][0]) # Output of hidden neuron A11 -> O11
  parameters['W2'][1][0] += (0.001 * 2 * (y-y_hat)*A1[1][0]) # Output of hidden neuron A12 -> O12
  parameters['b2'][0][0] += (0.001 * 2 * (y-y_hat))

  # Hidden Layer
  parameters['W1'][0][0] += (0.001 * 2 * (y-y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1] += (0.001 * 2 * (y-y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0] += (0.001 * 2 * (y-y_hat)*parameters['W2'][0][0])

  parameters['W1'][1][0] += (0.001 * 2 * (y-y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1] += (0.001 * 2 * (y-y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0] += (0.001 * 2 * (y-y_hat)*parameters['W2'][1][0])
```

```python
#epochs implementation
parameters = initialize_parameters([2,2,1])
epochs = 5

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):
    X = df[['cgpa','profile_score']].values[j].reshape(2,1) #shape(no. of features, no. of training samples)
    y = df[['lpa']].values[j][0]

    #parameter initialization
    y_hat, A1 = L_layer_forward(X, parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters, y, y_hat, A1, X)

    Loss.append((y-y_hat)**2)

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())
parameters
```

#Backpropagation Regression Intuition Using Keras

```python
import numpy as np
import pandas as pd
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa','profile_score','lpa'])
df
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2,activation='linear',input_dim=2)) # 2 Input Neurons and 2 Hidden Neurons
model.add(Dense(1,activation='linear')) #Output Neuron
```

```python
model.summary()  # 2*2(w)+2(b)=6, 2(w)+1(b)=3
```

```python
#weights that model is taking randomly
model.get_weights()
```

```python
#Updates weights according to my intuition
new_weights = [
    np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),
    np.array([0.,0.],dtype=np.float32),
    np.array([[0.1],[0.1]],dtype=np.float32),
    np.array([0.],dtype=np.float32)
]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error',optimizer=optimizer)
```

```python
X_train = df[['cgpa','profile_score']].values.astype(np.float32)
y_train = df['lpa'].values.astype(np.float32)
print(y_train.ndim)
print(X_train.shape[1])
```

```python
model.fit(X_train, y_train, epochs=75, verbose=1, batch_size=1)
```
