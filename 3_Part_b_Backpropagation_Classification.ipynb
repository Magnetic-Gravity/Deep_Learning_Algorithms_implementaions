# Backpropagation Using Classification Simple Intuition

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]], columns=['cgpa','profile_score','placed'])
df
```

```python
#layer_dims = [2(input_neuron), 2(hidden_neuron), 1(output_neuron)]
def initialize_parameters(layer_dims):
  np.random.seed(3)    # Fixes randomness so results are reproducible
  parameters = {}      # Store: [[W1, b1] ,[W2, b2]]
  L = len(layer_dims)

  for l in range(1,L): # We creates weight between layer0 and layer1
    parameters['W'+str(l)] = np.ones((layer_dims[l-1],layer_dims[l]))*0.1
    parameters['b'+str(l)] = np.zeros((layer_dims[l],1))
  return parameters
```

```python
#Activation Sigmoid
def sigmoid(Z):
  A = 1/(1+np.exp(-Z))
  return A
```

```python
# If First layer, A_prev = input features X and  If Hidden layer, A_prev = output of previous activation
def linear_forward(A_prev, W, b):
  Z = np.dot(W.T, A_prev) + b
  A = sigmoid(Z)
  return A
```

```python
#Forward Propagation
def L_layer_forward(X, parameters): # X-Input Features, Parameters-List containing all weights and biases
  A = X
  L = len(parameters) // 2     # Parameters [[W1,b1],[W2,b2]] which means 4//2 = 2

  for l in range(1, L+1):
    A_prev = A             # Save the previous layer’s output
    W1 = parameters['W' + str(l)]
    b1 = parameters['b' + str(l)]
    # print("A" + str(l-1) + " : ", A_prev)
    # print("W" + str(l) + " : ", W1)
    # print("b" + str(l) + " : ", b1)
    # print("--"*20)

    A = linear_forward(A_prev, W1, b1)
    # print("A" + str(l) + " : ", A)
    # print("**"*20)  # readability

  return A, A_prev   # A Output of Current, A_prev previous layer’s activation
```

```python
def update_parameters(parameters, y, y_hat, A1, X):
  # Output Layer
  parameters['W2'][0][0] += (0.001 * (y-y_hat)*A1[0][0]) # Output of hidden neuron A11 -> O11
  parameters['W2'][1][0] += (0.001 * (y-y_hat)*A1[1][0]) # Output of hidden neuron A12 -> O12
  parameters['b2'][0][0] += (0.001 * (y-y_hat))

  # Hidden Layer
  parameters['W1'][0][0] += (0.001 * (y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0])*X[0][0]) #A100 -> O11
  parameters['W1'][0][1] += (0.001 * (y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0])*X[1][0]) #A100 -> O11
  parameters['b1'][0][0] += (0.001 * (y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0]))

  parameters['W1'][1][0] += (0.001 * (y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0])*X[0][0])
  parameters['W1'][1][1] += (0.001 * (y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0])*X[1][0])
  parameters['b1'][1][0] += (0.001 * (y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0]))
```

```python
#epochs implementation
parameters = initialize_parameters([2,2,1])
epochs = 50

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):
    X = df[['cgpa','profile_score']].values[j].reshape(2,1) #shape(no. of features, no. of training samples)
    y = df[['placed']].values[j][0]

    #parameter initialization
    y_hat, A1 = L_layer_forward(X, parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters, y, y_hat, A1, X)

    Loss.append(-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())
parameters
```

#Backpropagation Classification Intuition Using Keras

```python
import numpy as np
import pandas as pd
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]], columns=['cgpa','profile_score','placed'])
df
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2,activation='sigmoid',input_dim=2)) # 2 Input Neurons and 2 Hidden Neurons
model.add(Dense(1,activation='sigmoid')) #Output Neuron
```

```python
model.summary()  # 2*2(w)+2(b)=6, 2(w)+1(b)=3
```

```python
#weights that model is taking randomly
model.get_weights()
```

```python
#Updates weights according to my intuition
new_weights = [
    np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),
    np.array([0.,0.],dtype=np.float32),
    np.array([[0.1],[0.1]],dtype=np.float32),
    np.array([0.],dtype=np.float32)
]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='binary_crossentropy',optimizer=optimizer)
```

```python
# model.fit(X_train, y_train, epochs=75, verbose=1, batch_size=1)
model.fit(df.iloc[:,0:-1].values, df['placed'].values, epochs=50, verbose=1, batch_size=1)
```
