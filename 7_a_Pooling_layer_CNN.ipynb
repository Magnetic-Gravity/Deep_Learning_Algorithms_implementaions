# Pooling (Downsampling):
- It is a crucial operation that reduces the spatial dimensions (width and height) of feature maps, minimizing computational load, reducing overfitting, and providing translation invariance. By summarizing small regions (e.g.,
pixels) of the input data, pooling helps the network focus on dominant features.

### Types of Pooling:
  - Max Pooling: Selects the highest value from each window of the feature map, commonly used to highlight the most dominant features (e.g., edges, strong patterns).
  - Average Pooling: Calculates the average value of the window, often used to produce a smoother output.
  - Global Pooling: An extreme form of pooling that reduces the entire feature map to a single value, typically used before the final classification layer.
  - Other Variations: Includes Min Pooling, Sum Pooling, or Lp pooling.

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,Conv2D,Flatten,MaxPooling2D
from keras.datasets import mnist
```

```python
(X_train,y_train), (X_test,y_train) = mnist.load_data()
```

```python
model=Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu',input_shape=(28,28,1)))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

### Usage Examples in CNN:
  - Dimensionality Reduction: Following convolutional layers to reduce image size and computation.
  - Translation Invariance: Allowing the model to recognize an object even if its position shifts slightly within the input image.
  - Overfitting Reduction: By shrinking data, the model generalizes better to new data.
  - Benefits and Disadvantages: Decreases training time, improves generalization, and reduces overfitting
