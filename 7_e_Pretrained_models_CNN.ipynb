Pre-trained models are trained on massive, general datasets, allowing developers to reuse them for specific tasks via fine-tuning. They save significant time and computational resources, acting as a starting point for specialized applications in NLP (e.g., BERT, GPT) and computer vision (e.g., ResNet, VGG).

- Benefits: Using pre-trained models significantly reduces training time, requires less data to achieve high accuracy, and lowers the barrier for AI adoption

```python
from tensorflow.keras.applications.resnet50 import ResNet50
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.resnet50 import preprocess_input, decode_predictions
import numpy as np
```

```python
model = ResNet50(weights='imagenet')
```

```python
img_path = '/content/chair.jpg'
img = image.load_img(img_path, target_size=(224, 224))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
```

```python
preds = model.predict(x)
print('Predicted : ', decode_predictions(preds, top=3)[0])
```
